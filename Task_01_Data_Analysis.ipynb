# Part 2: Data Cleaning and Preprocessing

In this part, we identify and handle missing values, check and remove duplicate records, inspect incorrect values, correct data types, and verify the cleaned dataset.

```python

# Check missing values and their percentages

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage (%)": missing_percentage
})

missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

print("Missing Values Analysis:")
display(missing_summary)
```

```python

# Visualize missing values

plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Values Distribution in Netflix Dataset")
plt.xlabel("Dataset Columns")
plt.ylabel("Dataset Records")

plt.tight_layout()
plt.show()
```

```python

# Identify duplicate records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)

# Remove duplicate records if they exist

if duplicate_count > 0:
    df = df.drop_duplicates().copy()
    print("Duplicate records removed successfully.")
else:
    print("No duplicate records found.")

# Verify duplicate removal

print("\nRemaining Duplicate Records:", df.duplicated().sum())

print("Updated Dataset Shape:", df.shape)
```

```python

# Check unique show identifiers

print("Total Show IDs:", df["show_id"].nunique())

print("Duplicate Show IDs:", df["show_id"].duplicated().sum())

print("\nShow Type Distribution:")
display(df["type"].value_counts())
```

```python

# Inspect current data types

print("Current Data Types:")
display(df.dtypes)

# Inspect date_added values

print("\nSample Date Values:")
display(df["date_added"].dropna().head(10))

# Inspect duration values

print("\nSample Duration Values:")
display(df["duration"].dropna().head(10))

# Inspect release years

print("\nRelease Year Summary:")
display(df["release_year"].describe())
```

```python

# Inspect column names

print("Current Column Names:")
print(df.columns.tolist())

# Check whether column names are duplicated

print("\nDuplicate Column Names:")
print(df.columns[df.columns.duplicated()].tolist())
```

```python

# Check unique values in categorical columns

print("Content Types:")
print(df["type"].value_counts(dropna=False))

print("\nRating Categories:")
print(df["rating"].value_counts(dropna=False))

print("\nRelease Year Range:")
print("Minimum:", df["release_year"].min())
print("Maximum:", df["release_year"].max())

print("\nDuration Values:")
print(df["duration"].value_counts(dropna=False).head(15))

print("\nCountry Sample:")
print(df["country"].dropna().unique()[:15])

print("\nListed Categories Sample:")
print(df["listed_in"].dropna().unique()[:15])
```

```python

# Inspect date values before conversion

parsed_dates = pd.to_datetime(
    df["date_added"].str.strip(),
    format="%B %d, %Y",
    errors="coerce"
)

print("Original Missing Dates:", df["date_added"].isna().sum())

print("Invalid or Unparseable Dates:",
      (df["date_added"].notna() & parsed_dates.isna()).sum())

print("\nSuccessfully Parsed Dates:")
display(parsed_dates.dropna().head(10))
```

```python

# Identify suspicious values in the rating column

suspicious_ratings = ["66 min", "74 min", "84 min"]

suspicious_rows = df[df["rating"].isin(suspicious_ratings)]

print("Records with Suspicious Rating Values:")
display(suspicious_rows)

print("\nNumber of Suspicious Records:", len(suspicious_rows))

# Inspect records with missing duration

print("\nRecords with Missing Duration:")
display(df[df["duration"].isna()])
```

```python

# Inspect all rating categories

print("Unique Rating Values:")
print(sorted(df["rating"].dropna().unique()))

print("\nRating Values Ending with 'min':")
print(
    df.loc[
        df["rating"].astype(str).str.endswith("min"),
        "rating"
    ].value_counts()
)
```

```python

# Convert date_added to datetime

df["date_added"] = pd.to_datetime(
    df["date_added"].str.strip(),
    format="%B %d, %Y",
    errors="coerce"
)

print("Updated Date Data Type:")
print(df["date_added"].dtype)

print("\nMissing Dates After Conversion:")
print(df["date_added"].isna().sum())

# Inspect duration data

print("\nDuration Data Type:")
print(df["duration"].dtype)

print("\nMissing Duration Values:")
print(df["duration"].isna().sum())
```

```python

# Create a backup of the original dataset

df_original = df.copy(deep=True)

print("Original dataset backup created successfully.")

print("Original Dataset Shape:", df_original.shape)
print("Working Dataset Shape:", df.shape)
```

```python
# Correct misplaced duration values

suspicious_ratings = ["66 min", "74 min", "84 min"]

mask = df["rating"].isin(suspicious_ratings)

df.loc[mask, "duration"] = df.loc[mask, "rating"]
df.loc[mask, "rating"] = np.nan

print("Corrected Records:")
display(df.loc[mask, ["show_id", "title", "rating", "duration"]])

print("\nMissing Rating Values:", df["rating"].isna().sum())
print("Missing Duration Values:", df["duration"].isna().sum())
```

```python
# Handle missing categorical values

columns_to_fill = ["director", "cast", "country", "rating"]

for column in columns_to_fill:
    df[column] = df[column].fillna("Unknown")

print("Missing Values After Filling:")
print(df[columns_to_fill].isnull().sum())
```

```python
# Verify data types after cleaning

print("Dataset Data Types:")
print(df.dtypes)

print("\nDate Added Data Type:")
print(df["date_added"].dtype)

print("\nRelease Year Data Type:")
print(df["release_year"].dtype)

print("\nDuration Data Type:")
print(df["duration"].dtype)
```

```python
# Final missing values analysis

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

display(missing_summary)
```

```python
# Verify cleaned dataset

print("Final Dataset Shape:", df.shape)
print("Total Duplicate Records:", df.duplicated().sum())

print("\nFinal Column Names:")
print(df.columns.tolist())

print("\nFinal Dataset Information:")
df.info()

print("\nCleaned Dataset Preview:")
display(df.head())
```

# Part 4: Data Visualization

This section is to represent the dataset through different visualizations, identify patterns, and communicate analytical findings clearly.

```python
# Prepare numerical features for visualization

df["movie_duration_minutes"] = np.where(
    df["type"] == "Movie",
    df["duration_value"],
    np.nan
)

df["tv_show_seasons"] = np.where(
    df["type"] == "TV Show",
    df["duration_value"],
    np.nan
)

print("Numerical Features Preview:")
display(
    df[
        [
            "type",
            "release_year",
            "movie_duration_minutes",
            "tv_show_seasons"
        ]
    ].head(10)
)

print("\nNumerical Features Summary:")
display(
    df[
        [
            "release_year",
            "movie_duration_minutes",
            "tv_show_seasons"
        ]
    ].describe().round(2)
)
```

```python
# Histogram: Distribution of Movie Durations

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df[df["type"] == "Movie"],
    x="movie_duration_minutes",
    bins=30,
    kde=True
)

plt.title("Distribution of Movie Durations")
plt.xlabel("Movie Duration (Minutes)")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()
```

```python
# Bar Chart: Top 10 Content Release Years

top_release_years = df["release_year"].value_counts().head(10).sort_index()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_release_years.index,
    y=top_release_years.values
)

plt.title("Top 10 Years by Number of Released Titles")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Pie Chart: Distribution of Movies and TV Shows

content_counts = df["type"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    content_counts.values,
    labels=content_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Movies and TV Shows")
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot: Release Year vs Movie Duration

movie_data = df[df["type"] == "Movie"]

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=movie_data,
    x="release_year",
    y="movie_duration_minutes",
    alpha=0.5
)

plt.title("Movie Release Year vs Movie Duration")
plt.xlabel("Release Year")
plt.ylabel("Movie Duration (Minutes)")
plt.tight_layout()
plt.show()
```

```python
# Line Chart: Netflix Content Added Over the Years

yearly_additions = (
    df["date_added_year"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_additions.index,
    yearly_additions.values,
    marker="o"
)

plt.title("Netflix Content Additions Over the Years")
plt.xlabel("Year Added to Netflix")
plt.ylabel("Number of Titles Added")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Box Plot: Distribution of Movie Durations

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df[df["type"] == "Movie"],
    x="movie_duration_minutes"
)

plt.title("Box Plot of Movie Durations")
plt.xlabel("Movie Duration (Minutes)")
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap: Numerical Features

correlation_data = df[
    [
        "release_year",
        "movie_duration_minutes",
        "tv_show_seasons"
    ]
]

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
# Count Plot: Number of Titles by Content Type

plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x="type"
)

plt.title("Count of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.tight_layout()
plt.show()
```

```python
# Pair Plot: Relationships Between Numerical Features

pairplot_data = df[
    [
        "release_year",
        "movie_duration_minutes",
        "tv_show_seasons"
    ]
].dropna()

sns.pairplot(
    pairplot_data,
    diag_kind="hist",
    corner=True
)

plt.suptitle(
    "Pair Plot of Numerical Features",
    y=1.02
)

plt.show()
```

```python
# Violin Plot: Movie Duration Distribution by Release Period

movie_violin_data = df[df["type"] == "Movie"].copy()

movie_violin_data["release_period"] = pd.cut(
    movie_violin_data["release_year"],
    bins=[1924, 2000, 2010, 2021],
    labels=["Before 2001", "2001–2010", "2011–2021"]
)

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=movie_violin_data,
    x="release_period",
    y="movie_duration_minutes"
)

plt.title("Movie Duration Distribution by Release Period")
plt.xlabel("Release Period")
plt.ylabel("Movie Duration (Minutes)")
plt.tight_layout()
plt.show()
```
